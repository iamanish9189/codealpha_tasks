```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'D:\archive\Iris.csv')
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
x =  df.drop(columns=['Species','Id'])
y = df['Species']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
x_train
```

```python
scaler = StandardScaler()
# fit the scaler to the train set, it will learn the parameters
scaler.fit(x_train)
# transform train and test sets
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train
```

```python
print(x.columns)
```

```python
np.round(x_train.describe(),1)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled,columns=x_test.columns)
```

```python
np.round(x_train_scaled.describe(),1)
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12,5))

# Before Scaling
ax1.set_title('Before Scaling')
sns.kdeplot(x_train['PetalLengthCm'], ax=ax1, label='PetalLengthCm')
sns.kdeplot(x_train['PetalWidthCm'], ax=ax1, label='PetalWidthCm')
sns.kdeplot(x_train['SepalLengthCm'], ax=ax1, label='SepalLengthCm')
sns.kdeplot(x_train['SepalWidthCm'], ax=ax1, label='SepalWidthCm')
ax1.legend()

# After Scaling
ax2.set_title('After Scaling')
sns.kdeplot(x_train_scaled['PetalLengthCm'], ax=ax2, label='PetalLengthCm')
sns.kdeplot(x_train_scaled['PetalWidthCm'], ax=ax2, label='PetalWidthCm')
sns.kdeplot(x_train_scaled['SepalLengthCm'], ax=ax2, label='SepalLengthCm')
sns.kdeplot(x_train_scaled['SepalWidthCm'], ax=ax2, label='SepalWidthCm')
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12,5))

# Before Scaling
ax1.set_title('Before Scaling')
sns.kdeplot(x_train['PetalLengthCm'], ax=ax1, label='PetalLengthCm')

# After Scaling
ax2.set_title('After Scaling')
sns.kdeplot(x_train_scaled['PetalLengthCm'], ax=ax2, label='PetalLengthCm')
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12,5))

# Before Scaling
ax1.set_title('Before Scaling')
sns.kdeplot(x_train['PetalWidthCm'], ax=ax1)

# After Scaling
ax2.set_title('After Scaling')
sns.kdeplot(x_train_scaled['PetalWidthCm'], ax=ax2)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
print("x_train:", x_train.shape)
print("x_test :", x_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
y_predict = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_predict)
print("Accuracy:", accuracy)
```

```python

result = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_predict
})

print(result)
```

```python
df['date'] = pd.to_datetime(df['date'],dayfirst=True)
df.info()
```

```python
df['month_int'] = df['date'].dt.month
df.head()
```

```python
import calendar
df['month'] = df['month_int'].apply(lambda x: calendar.month_abbr[x])
df.head()
```

```python
data = df.groupby(['month'])[['estimated unemployment rate','estimated employed',
                              'estimated labour participation rate']].mean()
data = pd.DataFrame(data).reset_index()
data
```

```python
import plotly.graph_objects as go

month = data.month
unemployment_rate = data['estimated unemployment rate']
labour_participation_rate = data['estimated labour participation rate']

fig = go.Figure()

fig.add_trace(go.Bar(x = month,y = unemployment_rate,name='estimated unemployment rate' ))
fig.add_trace(go.Bar(x = month, y = labour_participation_rate,name='estimated labour participation rate'))

fig.update_layout(title= 'Unemployment Rate and Labour Participation',
                  xaxis = {'categoryorder':'array','categoryarray':['jan','Feb',
               'Mar','Apr','May','Jun','July','Aug','sep','Oct']} )

fig.show()

```

```python
import plotly.express as px
```

```python
fig = px.bar(data,x='month',y = 'estimated employed',color='month',
             category_orders={'month':['jan','feb','mar','apr','May','Jun','Jul','Aug','Sept','Oct']},
             title='Estimated employed people from jan 2020 to oct 2020'
             )
fig.show()
```

```python
state =  df.groupby(['state'])[['estimated unemployment rate','estimated employed',
                                'estimated labour participation rate']].mean()
state = pd.DataFrame(state).reset_index()
```
